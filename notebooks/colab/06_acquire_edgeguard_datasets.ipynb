# EdgeGuard approved dataset acquisition
CPU-only thin wrapper. It lists each dataset's required artifacts or pinned generator identity. Runtime signed URLs, filenames, sizes and hashes are supplied only through environment variables after the corresponding terms/access decision. Missing gated artifacts remain blocked while other supplied artifacts continue. It never queues SMIYC or the frozen full Lost & Found holdout.

```python
import re
import subprocess
import sys
from pathlib import Path

from google.colab import drive

REPOSITORY = "https://github.com/emrealmaoglu/edgeguard-road.git"
EDGEGUARD_EXPECTED_COMMIT = "REPLACE_WITH_REVIEWED_EG_SEG_002_COMMIT_SHA"
PROJECT_ROOT = Path("/content/edgeguard-road")
if not re.fullmatch(r"[0-9a-f]{40}", EDGEGUARD_EXPECTED_COMMIT):
    raise ValueError("Enter the exact reviewed commit SHA")
subprocess.run(
    ["git", "clone", "--filter=blob:none", "--no-checkout", REPOSITORY, str(PROJECT_ROOT)],
    check=True,
)
subprocess.run(
    ["git", "-C", str(PROJECT_ROOT), "checkout", "--detach", EDGEGUARD_EXPECTED_COMMIT], check=True
)
subprocess.run([sys.executable, "-m", "pip", "install", "-e", str(PROJECT_ROOT)], check=True)
drive.mount("/content/drive")
```

```python
QUEUE = PROJECT_ROOT / "configs/dataset/acquisition_queue.yaml"
DRIVE_ROOT = Path("/content/drive/MyDrive/EdgeGuard")
subprocess.run(
    [
        sys.executable,
        str(PROJECT_ROOT / "scripts/acquire_edgeguard_datasets.py"),
        "--queue-config",
        str(QUEUE),
        "--drive-root",
        str(DRIVE_ROOT),
        "--list",
    ],
    check=True,
)
```
